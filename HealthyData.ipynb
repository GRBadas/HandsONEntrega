```python
# Lista das colunas que queremos analisar
colunas = ["cs_sexo", "cs_raca", "cs_escol_n", "evolucao", "nu_ano"]

# Função para calcular a porcentagem dos valores únicos
def calcular_percentuais(dataframe, colunas):
    resultados = {}
    for coluna in colunas:
        if coluna in dataframe.columns:
            percentuais = (
                dataframe[coluna]
                .value_counts(dropna=False, normalize=True) * 100
            ).round(5)
            resultados[coluna] = percentuais
    return resultados

# Aplicar a função
resultados = calcular_percentuais(df, colunas)

# Exibir os resultados
for coluna, valores in resultados.items():
    print(f"\nDistribuição percentual da coluna: {coluna}")
    print(valores)
```

```python
from scipy.stats import chi2_contingency

# Criar tabela de contingência
tabela_sexo = pd.crosstab(df['cs_sexo'], df['evolucao'])
chi2, p_value, dof, expected = chi2_contingency(tabela_sexo)

print(f"Teste Qui-Quadrado - Sexo vs Evolução:")
print(f"Chi² = {chi2:.4f}, p-value = {p_value:.4f}")

print(tabela_sexo)
```

Primeiro teste de hipotése, foi a hipotése o sexo possui uma inflência na evoluão do quadro do paciente, com um valor p < 0,005 podemos rejeitar a hipotése nula, ou seja não é possível, a hipotése de que o sexo NÃO tem influência na evolução do quadro. Entre os entrevistados 88,2% dos homens permaneceram vivo entre mulheres esse número cai para 84,5%. Quando observado o índice de mortes por Aids o número de mulheres é de 8,4% e o de homens é 6,4%. Quando consideradas mortes por qualquer comorbidade o número de mulheres é de 10% e o de homens de 7,7%

```python

# Preencher NaN ou valores faltantes
df['cs_raca'] = df['cs_raca'].fillna('Ignorado')

tabela_raca = pd.crosstab(df['cs_raca'], df['evolucao'])
chi2, p_value, dof, expected = chi2_contingency(tabela_raca)

print(f"Teste Qui-Quadrado - Raça vs Evolução:")
print(f"Chi² = {chi2:.4f}, p-value = {p_value:.4f}")
print(tabela_raca)
```

Segundo teste de hipotése, foi para avaliar se a raça possui uma influência na evoluão do quadro do paciente, com um valor p < 0,005 podemos rejeitar a hipotése nula, ou seja  é possível rejeitar a hipotése de que a raça NÃO tem influência na evolução do quadro. No dados Brancos possuem a maior porcentagem de vivos com o número de 90%,  enquanto Pretos e Pardos tem 86,6 e 87,1 respectivamente. Quando consideramos os mortos por aids os pretos lideram entre os não ignorados com 7,9% e pardos com 7,4 %. Indígenas e Amarelos estão entre esses dois extremos em ambas as taxas porém é interessante observa que entre indígenas não foram registradas mortes por outras comorbidades, apenas por aids.

```python
from scipy.stats import ttest_ind
import numpy as np

# Converter idade para numérico (tratando vírgula como decimal)
df['nu_idade_n'] = df['nu_idade_n'].str.replace(',', '.').astype(float)

# Separar idades por grupo de evolução
idades_vivos = df[df['evolucao'] == 'Vivo']['nu_idade_n'].dropna()
idades_obitos = df[df['evolucao'].str.contains('Óbito', na=False)]['nu_idade_n'].dropna()

t_stat, p_value = ttest_ind(idades_vivos, idades_obitos, equal_var=False)

print(f"Teste t - Idade vs Evolução:")
print(f"t = {t_stat:.4f}, p-value = {p_value:.4f}")
print(f"Idade média - Vivos: {np.mean(idades_vivos):.2f}")
print(f"Idade média - Óbitos: {np.mean(idades_obitos):.2f}")
```

O último teste de hipotése para medir influência da idade no quadro dos pacientes, de novo com o valor de p muito baixo, permite que se rejeite a hipotése nula, logo é possivel afirmar que a idade influência no quadro do paciente. Com a média de idade das pessoas que vieram a óbito mais alta que 7 anos da média de idade dos vivos indicando que pessoas acima dos 40 anos podem correr mais riscos de irem a óbito.

Concluindo, a influência nos quadros desse 3 fatores, idade, sexo e raça, é possível pensar que futuras politicas publicas e de conscientização deve forcar nesses grupos, mulheres, pretos e pardos e pessoa mais velhar atentando para as prvençoes e tratamentos cabíveis visto também que os números mostram que hoje muitas pessoas sobrevivem com a doença.
